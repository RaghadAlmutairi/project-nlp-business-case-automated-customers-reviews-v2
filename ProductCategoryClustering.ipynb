```python
# Check for missing values in the dataset
print("Missing values per column:")
print(df.isnull().sum())
```

```python
# List of columns to drop
columns_to_drop = [
    'id', 'dateAdded', 'dateUpdated', 'name', 'asins', 'brand', 'reviews.id',
    'primaryCategories', 'imageURLs', 'keys', 'manufacturer',
    'manufacturerNumber', 'reviews.date', 'reviews.dateAdded',
    'reviews.dateSeen', 'reviews.sourceURLs', 'sourceURLs'
]

# Drop the unnecessary columns
df = df.drop(columns=columns_to_drop)

# Verify the new shape and the columns
print(f"Shape after dropping unnecessary columns: {df.shape}")
df.head()
```

```python
# Fill remaining missing values with appropriate defaults
df['reviews.doRecommend'] = df['reviews.doRecommend'].fillna('No recommendation')
df['reviews.numHelpful'] = df['reviews.numHelpful'].fillna(0)
df['reviews.title'] = df['reviews.title'].fillna('No title')
df['reviews.username'] = df['reviews.username'].fillna('Anonymous')
df['reviews.didPurchase'] = df['reviews.didPurchase'].fillna('No')

# Final check
print("Final check for missing values:")
print(df.isnull().sum())
```

```python
# Check unique categories
df['categories'].unique()  # View all unique category entries
```

```python
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sentence_transformers import SentenceTransformer
```

```python
# Use a pre-trained Sentence-BERT model to encode the categories into embeddings
model = SentenceTransformer('all-MiniLM-L6-v2')
# Generate embeddings for the categories column
embeddings = model.encode(df['categories'].tolist(), show_progress_bar=True)
n_clusters = 4
```

```python
# K-means clustering on the embeddings
kmeans = KMeans(n_clusters=n_clusters, random_state=42)
df['cluster'] = kmeans.fit_predict(embeddings)
```

```python
# Visualize the clusters in 2D
plt.figure(figsize=(10, 8))
plt.scatter(embeddings[:, 0], embeddings[:, 1], c=df['cluster'], cmap='viridis', s=50)
plt.title('Product Category Clusters')
plt.xlabel(' Component 1')
plt.ylabel(' Component 2')
plt.colorbar(label='Cluster ID')
plt.show()
```

```python
# We can look at the categories within each cluster to get an idea of the themes

for i in range(n_clusters):
    print(f"Cluster {i}:")
    cluster_categories = df[df['cluster'] == i]['categories'].tolist()
    print(cluster_categories[:10])  # Print first 10 categories in this cluster
    print()
```

```python
# After inspecting the clusters, manually assign meaningful labels based on the themes

cluster_labels = {
    0: 'Ebook Readers',
    1: 'Batteries',
    2: 'Accessories',
    3: 'Non-Electronics',
    4: 'Computers'
}

df['meta_category'] = df['cluster'].map(cluster_labels)
```

```python
# Check the first few entries with their assigned meta-category
print(df[['categories', 'meta_category']].head())
```

```python
# Import Libraries
from sentence_transformers import SentenceTransformer
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
```

```python
# Generate Embeddings Using Sentence-BERT
# Using a transformer model to generate meaningful embeddings for product categories
model = SentenceTransformer('all-MiniLM-L6-v2')
embeddings = model.encode(df['categories'].tolist(), show_progress_bar=True)

# Scale the Embeddings (Optional but recommended for clustering)
scaler = StandardScaler()
embeddings_scaled = scaler.fit_transform(embeddings)

# Apply K-Means Clustering
# Elbow Method for determining optimal clusters (if not already decided)
inertia = []
silhouette_scores = []
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(embeddings_scaled)
    inertia.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(embeddings_scaled, kmeans.labels_))

# Plot the Elbow Method
plt.figure(figsize=(10, 6))
plt.plot(range(2, 11), inertia, marker='o')
plt.title('Elbow Method for Optimal Clusters')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.show()
```

```python
# Plot Silhouette Scores to assess cluster quality
plt.figure(figsize=(10, 6))
plt.plot(range(2, 11), silhouette_scores, marker='o')
plt.title('Silhouette Scores for Clustering')
plt.xlabel('Number of clusters')
plt.ylabel('Silhouette Score')
plt.show()
```

```python
# Based on the Elbow and Silhouette plots, choose the optimal number of clusters (e.g., 5)
n_clusters = 4 # Adjust this based on your analysis

# Apply K-Means with the chosen number of clusters
kmeans = KMeans(n_clusters=n_clusters, random_state=42)
df['cluster'] = kmeans.fit_predict(embeddings_scaled)

# Visualize Clusters Using PCA (Dimensionality Reduction)
pca = PCA(n_components=2)
reduced_embeddings = pca.fit_transform(embeddings_scaled)

plt.figure(figsize=(10, 8))
plt.scatter(reduced_embeddings[:, 0], reduced_embeddings[:, 1], c=df['cluster'], cmap='viridis', s=50)
plt.title('Product Category Clusters (k={})'.format(n_clusters), fontsize=16)
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.colorbar(label='Cluster ID')
plt.show()
```

```python
# Analyze the Clusters and Assign Meaningful Labels
# Inspect the first few categories in each cluster to determine labels
cluster_labels = {}
for i in range(n_clusters):
    print(f"\nCluster {i}:")
    cluster_categories = df[df['cluster'] == i]['categories'].tolist()
    # Print the first 10 categories for this cluster
    print(cluster_categories[:10])
```

```python
# Assign the final labels based on the analysis above
cluster_labels = {
    0: "Batteries & Household Electronics",
    1: "Tablets & eBook Readers",
    2: "Smart Home Devices & Audio",
    3: "Tablets & E-readers"
}

df['meta_category'] = df['cluster'].map(cluster_labels)

# Check the resulting meta-categories
print("\nFirst few entries with their meta-category:")
df[['categories', 'meta_category']].head()
```
