```python
# Map star ratings to sentiment classes
def map_star_rating_to_sentiment(rating):
    if rating <= 2:
        return 'Negative'
    elif rating == 3:
        return 'Neutral'
    elif rating >= 4:
        return 'Positive'

# Apply the mapping
df['sentiment'] = df['reviews.rating'].apply(map_star_rating_to_sentiment)

# Verify the new sentiment column
df.head()
```

```python
import re
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

def clean_text(text):
    if not isinstance(text, str):
        return ""
    # Remove special characters and digits
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    text = text.lower()
    words = text.split()
    words = [word for word in words if word not in ENGLISH_STOP_WORDS]
    return ' '.join(words)
```

```python
df['cleaned_reviews'] = df['reviews.text'].apply(clean_text)
df[['reviews.text', 'cleaned_reviews']].head()
```

```python
import pandas as pd
from sklearn.utils.class_weight import compute_class_weight
from sklearn.model_selection import train_test_split
from imblearn.over_sampling import SMOTE
from transformers import BertForSequenceClassification, BertTokenizer, Trainer, TrainingArguments
from sklearn.metrics import accuracy_score, precision_recall_fscore_support
import torch
from transformers import EarlyStoppingCallback
```

```python
# Map sentiment to numeric labels (Positive: 2, Neutral: 1, Negative: 0)
label2id = {'Negative': 0, 'Neutral': 1, 'Positive': 2}
id2label = {v: k for k, v in label2id.items()}
df['label'] = df['sentiment'].map(label2id)

# Split data into train and test sets
train_df, test_df = train_test_split(df[['cleaned_reviews', 'label']], test_size=0.2, stratify=df['label'], random_state=42)
```

```python
# Convert to Hugging Face Dataset
from datasets import Dataset
train_dataset = Dataset.from_pandas(train_df.reset_index(drop=True))
test_dataset = Dataset.from_pandas(test_df.reset_index(drop=True))

# Tokenize the data using BERT tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')

def tokenize(example):
    return tokenizer(example['cleaned_reviews'], padding='max_length', truncation=True, max_length=256)

train_dataset = train_dataset.map(tokenize, batched=True)
test_dataset = test_dataset.map(tokenize, batched=True)
```

```python
# Set format for PyTorch
train_dataset.set_format(type='torch', columns=['input_ids', 'attention_mask', 'label'])
test_dataset.set_format(type='torch', columns=['input_ids', 'attention_mask', 'label'])

# Load the model with class weights in the loss function
model = BertForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=3, id2label=id2label, label2id=label2id)
model.to('cuda' if torch.cuda.is_available() else 'cpu')

```

```python
import torch
from transformers import BertForSequenceClassification, BertTokenizer, Trainer, TrainingArguments
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
from transformers import EarlyStoppingCallback

# Ensure 'label' column is of type int
df['label'] = df['label'].astype(int)

# Calculate class weights for the imbalanced dataset
class_weights = compute_class_weight(
    'balanced',
    classes=np.array([0, 1, 2]),  # Convert the list to numpy array
    y=df['label'].values  # Use the target values
)

# Convert class weights to a tensor and move it to the appropriate device (CPU or GPU)
class_weights = torch.tensor(class_weights, dtype=torch.float).to('cuda' if torch.cuda.is_available() else 'cpu')

# Load the BERT model and tokenizer
model = BertForSequenceClassification.from_pretrained('bert-base-uncased', num_labels=3)
```

```python
# Define the custom compute_loss method inside the model
def compute_loss(model, inputs, return_outputs=False):
    """
    Custom loss function that includes class weights.
    """
    labels = inputs.get("labels")
    outputs = model(**inputs)
    logits = outputs.get("logits")

    # Use CrossEntropyLoss with class weights
    loss_fct = torch.nn.CrossEntropyLoss(weight=class_weights)
    loss = loss_fct(logits, labels)

    return (loss, outputs) if return_outputs else loss

# Override the model's compute_loss method
model.compute_loss = compute_loss
```

```python
# Define training arguments
training_args = TrainingArguments(
    output_dir="./sentiment_model",
    evaluation_strategy="epoch",
    save_strategy="epoch",
    load_best_model_at_end=True,
    metric_for_best_model="eval_f1",
    per_device_train_batch_size=16,
    per_device_eval_batch_size=16,
    num_train_epochs=5,
    learning_rate=5e-6,  # Lower learning rate
    weight_decay=0.01,
    warmup_steps=500,  # Learning rate warm-up
    logging_dir="./logs",
    logging_steps=50,
    report_to="none",  # Disable WandB logging
)

# Early stopping callback to avoid overfitting
early_stopping = EarlyStoppingCallback(early_stopping_patience=2)
```

```python
# Evaluation Metrics (precision, recall, f1, accuracy)
def compute_metrics(pred):
    labels = pred.label_ids
    preds = np.argmax(pred.predictions, axis=1)
    # Compute precision, recall, f1
    precision, recall, f1, _ = precision_recall_fscore_support(labels, preds, average='weighted')
    accuracy = accuracy_score(labels, preds)

    return {
        'eval_accuracy': accuracy,
        'eval_precision': precision,
        'eval_recall': recall,
        'eval_f1': f1
    }
```

```python
# Initialize the Trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=compute_metrics,  # Define your compute_metrics function
    callbacks=[early_stopping]
)

# Train the model
trainer.train()
```

```python
# Evaluate the model
results = trainer.evaluate()
print("Final Evaluation Results:")
results
```

```python
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

cm = confusion_matrix(test_df['label'], np.argmax(trainer.predict(test_dataset).predictions, axis=1))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=list(label2id.keys()), yticklabels=list(label2id.keys()))
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```
